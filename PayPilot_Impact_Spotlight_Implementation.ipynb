## Section 8: Payment and Revenue Allocation Logic

### Mathematical Model & Ledger Guarantees
Consider a client approving a milestone of **$1,000.00 USD** (100,000 cents):
1. **Base Milestone Amount:** $1,000.00
2. **Platform Fee (5%):**
   $$\text{Platform Fee} = \$1,000.00 \times 0.05 = \$50.00\text{ (5,000 cents)}$$
3. **Total Client Charge:** $1,050.00
4. **Freelancer Payout:** **$1,000.00 (100% intact, 0% deduction)**
5. **Impact Fund Allocation (30% of platform fee):**
   $$\text{Impact Allocation} = \$50.00 \times 0.30 = \$15.00\text{ (1,500 cents)}$$
6. **Platform Retained Revenue (70% of platform fee):**
   $$\text{Platform Retained} = \$50.00 \times 0.70 = \$35.00\text{ (3,500 cents)}$$

### Spotlight Cross-Funding Allocation
When a user purchases the **Featured Spotlight ($49.00 USD / 4,900 cents)**:
1. **Spotlight Gross Revenue:** $49.00
2. **Impact Fund Allocation (20% of Spotlight revenue):**
   $$\text{Spotlight Impact Allocation} = \$49.00 \times 0.20 = \$9.80\text{ (980 cents)}$$
3. **Platform Retained Spotlight Revenue (80%):**
   $$\text{Platform Retained} = \$49.00 \times 0.80 = \$39.20\text{ (3,920 cents)}$$

````python
def test_revenue_allocation_math(milestone_amount_usd: float, spotlight_price_usd: float):
    # Milestone Fee Allocation
    base_cents = int(round(milestone_amount_usd * 100))
    platform_fee_cents = int(round(base_cents * 0.05))
    impact_cents = int(round(platform_fee_cents * 0.30))
    platform_retained_cents = platform_fee_cents - impact_cents
    freelancer_payout_cents = base_cents  # 100% intact

    # Spotlight Allocation
    spotlight_cents = int(round(spotlight_price_usd * 100))
    spotlight_impact_cents = int(round(spotlight_cents * 0.20))
    spotlight_retained_cents = spotlight_cents - spotlight_impact_cents

    print(f"=== REVENUE ALLOCATION TEST (Milestone: ${milestone_amount_usd:.2f}) ===")
    print(f"1. Client Total Payment:     ${(base_cents + platform_fee_cents)/100:.2f}")
    print(f"2. Freelancer Guaranteed:   ${freelancer_payout_cents/100:.2f} (Deduction: $0.00)")
    print(f"3. Platform 5% Fee:          ${platform_fee_cents/100:.2f}")
    print(f"   -> Retained (70%):        ${platform_retained_cents/100:.2f}")
    print(f"   -> Impact Fund (30%):     ${impact_cents/100:.2f}")
    print(f"\n=== SPOTLIGHT ALLOCATION TEST (Price: ${spotlight_price_usd:.2f}) ===")
    print(f"1. Spotlight Gross:          ${spotlight_cents/100:.2f}")
    print(f"   -> Retained (80%):        ${spotlight_retained_cents/100:.2f}")
    print(f"   -> Impact Fund (20%):     ${spotlight_impact_cents/100:.2f}")

    assert freelancer_payout_cents == base_cents, "Freelancer earnings altered!"
    assert impact_cents == 1500, f"Expected 1500 cents, got {impact_cents}"
    assert spotlight_impact_cents == 980, f"Expected 980 cents, got {spotlight_impact_cents}"
    print("\nALL MATHEMATICAL LEDGER INVARIANTS SATISFIED.")

test_revenue_allocation_math(1000.0, 49.0)
````

## Section 9: PayPilot Impact Implementation

The PayPilot Impact subsystem manages social contributions, verified non-profit partner initiatives, and automated reputation badges:
- **Solvency Control:** Outward disbursements require a strict check against available unallocated funds:
  $$\text{Available Balance} = \sum \text{Inward Allocations} - \sum \text{Disbursed Capital} \ge \text{Disbursement Request}$$
  If the requested disbursement exceeds the solvent balance, an `HTTP 400 Insufficient available impact funds` is raised and the transaction is aborted.
- **Badge Award Engine:** Whenever an allocation is recorded for a client or freelancer, the engine sums their lifetime contributions and awards qualifying badges (`Starter` at $10, `Supporter` at $50, `Champion` at $250, `Leader` at $1,000).
